```python
import sympy as sp
#a)
# PUNTO 1.A

# Variables
x, y = sp.symbols('x y')
a0, a1, a2, b0, b1, b2 = sp.symbols('a0 a1 a2 b0 b1 b2')

# P2(x) y G2(y)
h = (a0 + a1*x + a2*x**2)*(b0 + b1*y + b2*y**2)

# bases monomiales
base_x_monom = [1, x, x**2]
base_y_monom = [1, y, y**2]

# C^ij
coeffs_monom = []
for i in range(3):
    for j in range(3):
        coeffs_monom.append(sp.expand(h).coeff(x, i).coeff(y, j))

coeffs_monom = sp.Matrix(coeffs_monom)

print("Coeficientes en base monomial (columna 9x1):\n")
sp.pprint(coeffs_monom)
```

```python
# PUNTO 1.B - Adaptado para mostrar la expansión explícita
#b)
x, y = sp.symbols('x y')

# Bases
basis_x = [1, x, x**2]   # e1=1, e2=x, e3=x^2
basis_y = [1, y, y**2]   # f1=1, f2=y, f3=y^2

# Polinomios dados
pP = x**2 + x + 3
pG = y + 1

# Tensor (producto exterior)
T = sp.expand(pP * pG)
print("Polinomio tensor T(x,y) = p^P(x) ⊗ p^G(y):")
sp.pprint(T)
print()

# Calcular componentes c^{ij}
C = sp.Matrix([[sp.expand(T).coeff(x, i).coeff(y, j) for j in range(3)] for i in range(3)])
print("Matriz de componentes c^{ij} (filas= e1,e2,e3 ; columnas= f1,f2,f3):")
sp.pprint(C)
print()

# Expansión explícita
print("Expansión en la base tensorial:")
terms = []
for i, ex in enumerate(basis_x, start=1):
    for j, ey in enumerate(basis_y, start=1):
        coeff = C[i-1, j-1]
        if coeff != 0:
            terms.append(f"{coeff}·(e{i}⊗f{j}) = {coeff}·({ex}⊗{ey})")

for t in terms:
    print(t)
```

```python
#c)
# Variable
x = sp.Symbol('x')

# Polinomio dado
p = x**2 + x + 3

# Polinomios de Legendre
P0 = 1
P1 = x
P2 = sp.Rational(1,2)*(3*x**2 - 1)

# ====== MÉTODO A: Comparación de coeficientes ======
# Expresamos p como α0*P0 + α1*P1 + α2*P2
a0, a1, a2 = sp.symbols('a0 a1 a2')
expr = a0*P0 + a1*P1 + a2*P2

# Igualamos a p(x)
eq = sp.expand(expr - p)

# Coeficientes en la base monomial [1, x, x^2]
eqs = [
    sp.Eq(sp.expand(eq).coeff(x,0), 0),  # término independiente
    sp.Eq(sp.expand(eq).coeff(x,1), 0),  # coef x
    sp.Eq(sp.expand(eq).coeff(x,2), 0)   # coef x^2
]

sol_A = sp.solve(eqs, [a0, a1, a2])

# ====== MÉTODO B: Cambio de base ======
# Matriz de cambio base monomial -> Legendre
# Cada columna es la representación de Pi en la base [1, x, x^2]
basis_monom = [1, x, x**2]
basis_legendre = [P0, P1, P2]

M = sp.Matrix([
    [sp.expand(P).coeff(x,0), sp.expand(P).coeff(x,1), sp.expand(P).coeff(x,2)]
    for P in basis_legendre
]).T  # transponemos para que cada columna sea Pi

# Vector de coeficientes de p en la base monomial [1, x, x^2]
p_vec = sp.Matrix([p.coeff(x,0), p.coeff(x,1), p.coeff(x,2)])

# Resolvemos M * [a0, a1, a2]^T = p_vec
a_vec = M.LUsolve(p_vec)

# ====== Resultados ======
print("Coeficientes método A:", sol_A)
print("Coeficientes método B:", a_vec)
```

```python
#d)
from sympy import symbols, legendre, integrate, Matrix

# Definir las variables
x, y = symbols('x y')

# Polinomios originales
p_P = x**2 + x + 3  # p^P(x) = x² + 2x + 3
p_G = y + 1           # p^G(y) = y + 1

# Producto tensorial
p_tensor = p_P * p_G
print(f"Producto tensorial: p^{{P⊗G}}(x,y) = {p_tensor}")
print()

# Bases de polinomios de Legendre (grado ≤ 2)
# Para P_2(x)
P0_x = legendre(0, x)  # P₀(x) = 1
P1_x = legendre(1, x)  # P₁(x) = x
P2_x = legendre(2, x)  # P₂(x) = (3x² - 1)/2

# Para G_2(y)
P0_y = legendre(0, y)  # P₀(y) = 1
P1_y = legendre(1, y)  # P₁(y) = y
P2_y = legendre(2, y)  # P₂(y) = (3y² - 1)/2

print("Bases de polinomios de Legendre:")
print(f"Para P_2(x): P₀(x) = {P0_x}, P₁(x) = {P1_x}, P₂(x) = {P2_x}")
print(f"Para G_2(y): P₀(y) = {P0_y}, P₁(y) = {P1_y}, P₂(y) = {P2_y}")
print()

# Normas al cuadrado de los polinomios de Legendre
norm_P0 = 2    # ∫₋₁¹ P₀(x)² dx = 2
norm_P1 = 2/3  # ∫₋₁¹ P₁(x)² dx = 2/3
norm_P2 = 2/5  # ∫₋₁¹ P₂(x)² dx = 2/5

# Función para calcular coeficientes usando ortogonalidad
def legendre_coeff(f, P_i, norm, var):
    """Calcula el coeficiente de Legendre ⟨f|P_i⟩/⟨P_i|P_i⟩"""
    numerator = integrate(f * P_i, (var, -1, 1))
    return numerator / norm

# Calcular las componentes c^{ij} usando la ortogonalidad de la base tensorial
# c^{ij} = ⟨p^{P⊗G}(x,y) | P_i(x) ⊗ P_j(y)⟩ / (⟨P_i|P_i⟩⟨P_j|P_j⟩)

c_matrix = sp.Matrix.zeros(3, 3)

print("Cálculo de componentes c^{ij}:")
print("=" * 50)

for i in range(3):
    for j in range(3):
        # Seleccionar los polinomios de Legendre apropiados
        if i == 0:
            P_i = P0_x
            norm_i = norm_P0
        elif i == 1:
            P_i = P1_x
            norm_i = norm_P1
        else:
            P_i = P2_x
            norm_i = norm_P2

        if j == 0:
            P_j = P0_y
            norm_j = norm_P0
        elif j == 1:
            P_j = P1_y
            norm_j = norm_P1
        else:
            P_j = P2_y
            norm_j = norm_P2

        # Función base tensorial
        basis_ij = P_i * P_j

        # Numerador: ⟨p^{P⊗G} | P_i ⊗ P_j⟩
        numerator = integrate(p_tensor * basis_ij, (x, -1, 1), (y, -1, 1))

        # Denominador: ⟨P_i|P_i⟩⟨P_j|P_j⟩
        denominator = norm_i * norm_j

        # Componente c^{ij}
        c_ij = numerator / denominator
        c_matrix[i, j] = c_ij

        print(f"c^{{{i}{j}}} = ∫∫ p(x,y)·P_{i}(x)P_{j}(y) dxdy / (⟨P_{i}|P_{i}⟩⟨P_{j}|P_{j}⟩)")
        print(f"     = {numerator} / ({norm_i}·{norm_j}) = {c_ij}")
        print()

print("Matriz de componentes c^{ij}:")
print("c =")
print(c_matrix)
print()

# Verificación: reconstruir el polinomio a partir de los coeficientes
print("Verificación - Reconstrucción del polinomio:")
reconstructed = 0
for i in range(3):
    for j in range(3):
        if i == 0:
            P_i = P0_x
        elif i == 1:
            P_i = P1_x
        else:
            P_i = P2_x

        if j == 0:
            P_j = P0_y
        elif j == 1:
            P_j = P1_y
        else:
            P_j = P2_y

        reconstructed += c_matrix[i, j] * P_i * P_j

print(f"Polinomio reconstruido: {sp.expand(reconstructed)}")
print(f"Polinomio original:     {sp.expand(p_tensor)}")
print("¿Coinciden?", sp.simplify(reconstructed - p_tensor) == 0)
print()

# Mostrar la expansión completa
print("Expansión en base de Legendre:")
print("p^{P⊗G}(x,y) = ")
for i in range(3):
    for j in range(3):
        if c_matrix[i, j] != 0:
            sign = "+" if c_matrix[i, j] >= 0 else ""
            print(f"{sign} {c_matrix[i, j]}·[P_{i}(x) ⊗ P_{j}(y)]")
```
